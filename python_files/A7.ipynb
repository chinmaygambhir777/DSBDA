```python
import nltk
from nltk.tokenize import *
from nltk.corpus import *
from nltk.stem import *
from nltk import pos_tag
import re 
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download('all')
```

```python
text="Hey! My name is Diya. I am a computer engineering student. I know Python,Java and C++. In my free time, I like listening music and reading books."
text=re.sub('[^A-Za-z]',' ',text)
print(text)
```

```python
tokens = word_tokenize(text)
print(tokens)
```

```python
pos = pos_tag(tokens)
print(pos)
```

```python
stop = set(stopwords.words('english'))
filtered_words = []
for word in tokens:
    if word.lower() not in stop:
        filtered_words.append(word)
print("Stop Words - ", stop)
print("\nfiltered_words",filtered_words)
```

```python
ps = PorterStemmer()
stemmed_words = []
for word in filtered_words:
    s = ps.stem(word)
    stemmed_words.append(s)
    print(word," -> ",s)
    
```

```python
lemmatizer = WordNetLemmatizer()
for word in filtered_words:
    print(word ," -> ", lemmatizer.lemmatize(word))
```

```python
documents = ["This is example of term Frequency. meow meow meow meow"
]

tf_idf = TfidfVectorizer()
tfidf_matrix = tf_idf.fit_transform(documents)
feature_names = tf_idf.get_feature_names_out()
print("TF-IDF Matrix: ", tfidf_matrix.toarray())
print("Features : ",feature_names)
```
